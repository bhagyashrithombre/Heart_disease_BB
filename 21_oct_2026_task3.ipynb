### Test model performance on different train-test splits by iteratively adjusting data distributions, evaluating shifts in accuracy, and ensuring stability across multiple randomized and stratified splits.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.base import clone
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("heart_disease.csv")

print("Dataset loaded successfully")
print("Dataset shape:", df.shape)

display(df.head())
print(df.columns.tolist())
```

```python
print(df.columns.tolist())

target_column = "target"

if target_column not in df.columns:
    raise ValueError(
        f"'{target_column}' not found. "
        f"Available columns: {df.columns.tolist()}"
    )

df = df.dropna(subset=[target_column]).copy()

X = df.drop(columns=[target_column])
y = df[target_column]

X = pd.get_dummies(X, drop_first=True, dtype=int)

if not pd.api.types.is_numeric_dtype(y):
    y = pd.factorize(y)[0]

X = X.reset_index(drop=True)
y = pd.Series(np.asarray(y).ravel()).reset_index(drop=True)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts())
```

```python
models = {
    "Logistic Regression": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        LogisticRegression(max_iter=2000, random_state=42)
    ),

    "KNN": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        KNeighborsClassifier(n_neighbors=5)
    ),

    "SVM": make_pipeline(
        SimpleImputer(strategy="median"),
        StandardScaler(),
        SVC(probability=True, random_state=42)
    ),

    "Decision Tree": make_pipeline(
        SimpleImputer(strategy="median"),
        DecisionTreeClassifier(random_state=42)
    ),

    "Random Forest": make_pipeline(
        SimpleImputer(strategy="median"),
        RandomForestClassifier(
            n_estimators=100,
            random_state=42
        )
    )
}

print("Models defined successfully")
```

```python
results = []

test_sizes = [0.30, 0.25, 0.20]
random_seeds = [11, 21, 42, 55, 75]

for test_size in test_sizes:
    for seed in random_seeds:

        for method in ["Random", "Stratified"]:

            stratify = y if method == "Stratified" else None

            if method == "Stratified" and y.value_counts().min() < 2:
                continue

            try:
                X_train, X_test, y_train, y_test = train_test_split(
                    X,
                    y,
                    test_size=test_size,
                    random_state=seed,
                    stratify=stratify
                )
            except ValueError as e:
                print("Split skipped:", e)
                continue

            for name, model in models.items():

                fitted_model = clone(model)
                fitted_model.fit(X_train, y_train)

                y_pred = fitted_model.predict(X_test)

                auc = np.nan

                if y.nunique() == 2:
                    try:
                        y_score = fitted_model.predict_proba(X_test)[:, 1]
                        auc = roc_auc_score(y_test, y_score)
                    except (ValueError, IndexError):
                        pass

                results.append({
                    "Model": name,
                    "Train Size": round(1 - test_size, 2),
                    "Test Size": test_size,
                    "Split Method": method,
                    "Random Seed": seed,
                    "Accuracy": accuracy_score(y_test, y_pred),
                    "Precision": precision_score(
                        y_test, y_pred,
                        average="weighted",
                        zero_division=0
                    ),
                    "Recall": recall_score(
                        y_test, y_pred,
                        average="weighted",
                        zero_division=0
                    ),
                    "F1 Score": f1_score(
                        y_test, y_pred,
                        average="weighted",
                        zero_division=0
                    ),
                    "ROC-AUC": auc
                })

results_df = pd.DataFrame(results)

display(results_df.head(10))
print("Total experiments:", len(results_df))
```

```python
summary = results_df.groupby("Model").agg(
    Mean_Accuracy=("Accuracy", "mean"),
    Std_Accuracy=("Accuracy", "std"),
    Mean_Precision=("Precision", "mean"),
    Mean_Recall=("Recall", "mean"),
    Mean_F1=("F1 Score", "mean"),
    Mean_ROC_AUC=("ROC-AUC", "mean")
).round(4)

display(summary.sort_values("Mean_Accuracy", ascending=False))
```

```python
plot_data = results_df.groupby(
    ["Model", "Test Size"]
)["Accuracy"].mean().reset_index()

for name in models:
    data = plot_data[plot_data["Model"] == name]

    plt.plot(
        data["Test Size"] * 100,
        data["Accuracy"],
        marker="o",
        label=name
    )

plt.xlabel("Test Set Size (%)")
plt.ylabel("Mean Accuracy")
plt.title("Model Stability Across Train-Test Splits")
plt.xticks([20, 25, 30])
plt.legend()
plt.grid(True)
plt.show()
```

```python
stability = results_df.groupby("Model")["Accuracy"].agg(
    ["mean", "std", "min", "max"]
).round(4)

stability["Accuracy_Range"] = (
    stability["max"] - stability["min"]
)

display(stability.sort_values("std", ascending=True))
```

### Perform adversarial testing by slightly modifying input features using perturbation techniques, manually designing worst-case adversarial samples, and quantifying their impact on model robustness.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

model.fit(X_train, y_train)

y_original = model.predict(X_test)

print("Baseline Accuracy:",
      round(accuracy_score(y_test, y_original), 4))

print("Baseline F1 Score:",
      round(f1_score(y_test, y_original,
                     average="weighted", zero_division=0), 4))
```

```python
continuous_features = [
    col for col in X_train.columns
    if pd.api.types.is_numeric_dtype(X_train[col])
    and X_train[col].nunique() > 10
]

if not continuous_features:
    raise ValueError(
        "No continuous numerical features found. "
        "Inspect X_train and choose appropriate features manually."
    )

print("Features selected for perturbation:")
print(continuous_features)

feature_scales = {}

for col in continuous_features:
    q1 = X_train[col].quantile(0.25)
    q3 = X_train[col].quantile(0.75)
    feature_scales[col] = q3 - q1

display(pd.DataFrame({
    "Feature": continuous_features,
    "Training IQR": [
        feature_scales[c] for c in continuous_features
    ]
}))
```

```python
X_perturbed = X_test.copy()

for col in continuous_features:
    shift = 0.05 * feature_scales[col]

    lower = X_train[col].quantile(0.01)
    upper = X_train[col].quantile(0.99)

    direction = np.where(
        np.arange(len(X_test)) % 2 == 0, 1, -1
    )

    X_perturbed[col] = np.clip(
        X_test[col].to_numpy() + direction * shift,
        lower,
        upper
    )

y_perturbed = model.predict(X_perturbed)

print("Original predictions:")
print(np.asarray(y_original)[:10])

print("\nPerturbed predictions:")
print(np.asarray(y_perturbed)[:10])
```

```python
def evaluate_predictions(y_true, y_pred, name):
    return {
        "Test Type": name,
        "Accuracy": accuracy_score(y_true, y_pred),
        "Precision": precision_score(
            y_true, y_pred, average="weighted", zero_division=0
        ),
        "Recall": recall_score(
            y_true, y_pred, average="weighted", zero_division=0
        ),
        "F1 Score": f1_score(
            y_true, y_pred, average="weighted", zero_division=0
        )
    }

baseline_metrics = evaluate_predictions(
    y_test, y_original, "Original"
)

perturbed_metrics = evaluate_predictions(
    y_test, y_perturbed, "Small Perturbation"
)

performance_comparison = pd.DataFrame([
    baseline_metrics,
    perturbed_metrics
])

display(performance_comparison.round(4))

flip_rate = np.mean(
    np.asarray(y_original) != np.asarray(y_perturbed)
)

print(f"Prediction flip rate: {flip_rate:.2%}")
print(
    f"Accuracy change: "
    f"{accuracy_score(y_test, y_perturbed) - accuracy_score(y_test, y_original):.4f}"
)
```

```python
X_worst = X_test.copy()

original_predictions = model.predict(X_test)
original_probabilities = model.predict_proba(X_test)

classes = model[-1].classes_

original_class_indices = np.array([
    np.where(classes == pred)[0][0]
    for pred in original_predictions
])

original_confidence = original_probabilities[
    np.arange(len(X_test)),
    original_class_indices
]

worst_confidence = original_confidence.copy()

for col in continuous_features:
    shift = 0.25 * feature_scales[col]

    lower = X_train[col].quantile(0.01)
    upper = X_train[col].quantile(0.99)

    for direction in [-1, 1]:
        candidate = X_test.copy()

        candidate[col] = np.clip(
            X_test[col] + direction * shift,
            lower,
            upper
        )

        probabilities = model.predict_proba(candidate)

        candidate_confidence = probabilities[
            np.arange(len(X_test)),
            original_class_indices
        ]

        update_mask = candidate_confidence < worst_confidence

        X_worst.loc[update_mask, col] = candidate.loc[
            update_mask, col
        ]

        worst_confidence[update_mask] = (
            candidate_confidence[update_mask]
        )

y_worst = model.predict(X_worst)

print("Worst-case accuracy:",
      round(accuracy_score(y_test, y_worst), 4))

print("Worst-case prediction flip rate:",
      f"{np.mean(y_worst != original_predictions):.2%}")

print("Average original confidence:",
      round(original_confidence.mean(), 4))

print("Average worst-case confidence:",
      round(worst_confidence.mean(), 4))
```

```python
worst_metrics = evaluate_predictions(
    y_test, y_worst, "Worst-Case Perturbation"
)

final_comparison = pd.DataFrame([
    baseline_metrics,
    perturbed_metrics,
    worst_metrics
])

display(final_comparison.round(4))

plt.figure(figsize=(9, 5))

for metric in ["Accuracy", "Precision", "Recall", "F1 Score"]:
    plt.plot(
        final_comparison["Test Type"],
        final_comparison[metric],
        marker="o",
        label=metric
    )

plt.title("Model Robustness Under Adversarial Perturbations")
plt.ylabel("Score")
plt.xlabel("Test Scenario")
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True)
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
sample_analysis = X_test.copy()

sample_analysis["Actual"] = np.asarray(y_test)
sample_analysis["Original Prediction"] = original_predictions
sample_analysis["Perturbed Prediction"] = y_perturbed
sample_analysis["Worst-Case Prediction"] = y_worst
sample_analysis["Original Confidence"] = original_confidence
sample_analysis["Worst-Case Confidence"] = worst_confidence

sample_analysis["Prediction Changed"] = (
    sample_analysis["Original Prediction"]
    != sample_analysis["Worst-Case Prediction"]
)

sample_analysis["Confidence Drop"] = (
    original_confidence - worst_confidence
)

display(
    sample_analysis.sort_values(
        "Confidence Drop", ascending=False
    ).head(10).round(4)
)
```

```python
baseline_accuracy = accuracy_score(y_test, y_original)
small_accuracy = accuracy_score(y_test, y_perturbed)
worst_accuracy = accuracy_score(y_test, y_worst)

print("===== ROBUSTNESS REPORT =====")
print(f"Baseline accuracy: {baseline_accuracy:.4f}")
print(f"Small-perturbation accuracy: {small_accuracy:.4f}")
print(f"Worst-case accuracy: {worst_accuracy:.4f}")

print(f"Small-perturbation accuracy change: "
      f"{small_accuracy - baseline_accuracy:+.4f}")

print(f"Worst-case accuracy change: "
      f"{worst_accuracy - baseline_accuracy:+.4f}")

print(f"Small-perturbation flip rate: "
      f"{np.mean(y_original != y_perturbed):.2%}")

print(f"Worst-case flip rate: "
      f"{np.mean(y_original != y_worst):.2%}")

print(f"Mean confidence reduction: "
      f"{np.mean(original_confidence - worst_confidence):.4f}")
```

### Check model performance on out-of-distribution samples by curating synthetic edge cases, analyzing model confidence levels, and assessing generalization ability beyond the training data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = make_pipeline(
    SimpleImputer(strategy="median"),
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

model.fit(X_train, y_train)

y_test_pred = model.predict(X_test)
test_probabilities = model.predict_proba(X_test)

baseline_accuracy = accuracy_score(y_test, y_test_pred)

print("Baseline test accuracy:", round(baseline_accuracy, 4))
print("Baseline precision:", round(
    precision_score(y_test, y_test_pred,
                    average="weighted", zero_division=0), 4
))
print("Baseline recall:", round(
    recall_score(y_test, y_test_pred,
                 average="weighted", zero_division=0), 4
))
print("Baseline F1-score:", round(
    f1_score(y_test, y_test_pred,
             average="weighted", zero_division=0), 4
))
```

```python
continuous_features = [
    col for col in X_train.columns
    if X_train[col].nunique() > 10
]

print("Continuous features:", continuous_features)

if not continuous_features:
    raise ValueError(
        "No continuous features identified. "
        "Inspect X_train and choose numerical features manually."
    )

feature_bounds = {}

for col in continuous_features:
    q01 = X_train[col].quantile(0.01)
    q25 = X_train[col].quantile(0.25)
    q75 = X_train[col].quantile(0.75)
    q99 = X_train[col].quantile(0.99)

    iqr = q75 - q25

    margin = 0.25 * iqr

    feature_bounds[col] = {
        "lower": q01 - margin,
        "upper": q99 + margin,
        "train_min": X_train[col].min(),
        "train_max": X_train[col].max()
    }

display(pd.DataFrame(feature_bounds).T.round(3))
```

```python
ood_samples = []
ood_metadata = []

for col in continuous_features:
    for edge in ["Low Edge", "High Edge"]:

        X_edge = X_test.copy()

        edge_value = (
            feature_bounds[col]["lower"]
            if edge == "Low Edge"
            else feature_bounds[col]["upper"]
        )

        X_edge[col] = edge_value

        for i in range(len(X_edge)):
            ood_samples.append(X_edge.iloc[i].to_dict())

            ood_metadata.append({
                "Original_Row": i,
                "Modified_Feature": col,
                "Edge_Type": edge,
                "Edge_Value": edge_value
            })

X_ood = pd.DataFrame(ood_samples, columns=X_train.columns)
ood_metadata_df = pd.DataFrame(ood_metadata)

print("Original test samples:", len(X_test))
print("Synthetic edge-case samples:", len(X_ood))

display(ood_metadata_df.head())
```

```python
ood_predictions = model.predict(X_ood)
ood_probabilities = model.predict_proba(X_ood)

ood_confidence = ood_probabilities.max(axis=1)

original_predictions_repeated = np.asarray(y_test_pred)[
    ood_metadata_df["Original_Row"].to_numpy()
]

ood_results = ood_metadata_df.copy()

ood_results["Original_Prediction"] = original_predictions_repeated
ood_results["OOD_Prediction"] = ood_predictions
ood_results["OOD_Confidence"] = ood_confidence

ood_results["Prediction_Changed"] = (
    ood_results["Original_Prediction"]
    != ood_results["OOD_Prediction"]
)

ood_results["Confidence_Change"] = (
    ood_confidence
    - test_probabilities.max(axis=1)[
        ood_metadata_df["Original_Row"].to_numpy()
    ]
)

display(
    ood_results.sort_values(
        "OOD_Confidence"
    ).head(15).round(4)
)
```

```python
baseline_confidence = test_probabilities.max(axis=1)

confidence_summary = pd.DataFrame({
    "Dataset": ["Original Test Data", "Synthetic OOD Data"],
    "Mean Confidence": [
        baseline_confidence.mean(),
        ood_confidence.mean()
    ],
    "Median Confidence": [
        np.median(baseline_confidence),
        np.median(ood_confidence)
    ],
    "Low Confidence Rate (< 0.60)": [
        np.mean(baseline_confidence < 0.60),
        np.mean(ood_confidence < 0.60)
    ]
})

display(confidence_summary.round(4))

plt.figure(figsize=(9, 5))

plt.hist(
    baseline_confidence,
    bins=15,
    alpha=0.6,
    label="Original Test Data"
)

plt.hist(
    ood_confidence,
    bins=15,
    alpha=0.6,
    label="Synthetic OOD Data"
)

plt.xlabel("Maximum Predicted Probability")
plt.ylabel("Number of Samples")
plt.title("Prediction Confidence: In-Distribution vs OOD")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
feature_impact = (
    ood_results
    .groupby(["Modified_Feature", "Edge_Type"])
    .agg(
        Mean_Confidence=("OOD_Confidence", "mean"),
        Prediction_Flip_Rate=("Prediction_Changed", "mean"),
        Mean_Confidence_Change=("Confidence_Change", "mean")
    )
    .reset_index()
)

feature_impact["Prediction_Flip_Rate"] *= 100

display(
    feature_impact.sort_values(
        "Prediction_Flip_Rate",
        ascending=False
    ).round(4)
)
```

```python
flip_plot = (
    ood_results.groupby("Modified_Feature")["Prediction_Changed"]
    .mean()
    .sort_values(ascending=False) * 100
)

plt.figure(figsize=(10, 5))
flip_plot.plot(kind="bar")
plt.ylabel("Prediction Flip Rate (%)")
plt.xlabel("Modified Feature")
plt.title("Feature-wise OOD Sensitivity")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
X_ood_check = X_ood.reset_index(drop=True)

outside_range = pd.DataFrame(index=X_ood_check.index)

for col in continuous_features:
    outside_range[col] = (
        (X_ood_check[col] < X_train[col].min()) |
        (X_ood_check[col] > X_train[col].max())
    )

print(
    "Percentage of OOD samples with at least one feature "
    "outside the training range:",
    round(outside_range.any(axis=1).mean() * 100, 2),
    "%"
)

print("\nOutside-range rate by feature:")
display(
    (outside_range.mean() * 100)
    .sort_values(ascending=False)
    .rename("Outside Training Range (%)")
    .to_frame()
    .round(2)
)
```

```python
flip_rate = ood_results["Prediction_Changed"].mean()

mean_confidence_drop = (
    baseline_confidence.mean() - ood_confidence.mean()
)

report = pd.DataFrame({
    "Metric": [
        "Baseline Test Accuracy",
        "Baseline Mean Confidence",
        "OOD Mean Confidence",
        "OOD Prediction Flip Rate",
        "Mean Confidence Reduction",
        "OOD Samples Outside Training Range"
    ],
    "Value": [
        accuracy_score(y_test, y_test_pred),
        baseline_confidence.mean(),
        ood_confidence.mean(),
        flip_rate,
        mean_confidence_drop,
        outside_range.any(axis=1).mean()
    ]
})

display(report.round(4))

print("\n===== OOD TESTING CONCLUSION =====")

if flip_rate > 0.20:
    print("The model shows substantial prediction sensitivity.")
else:
    print("The model shows relatively fewer prediction changes.")

if ood_confidence.mean() < baseline_confidence.mean():
    print("Average confidence decreased on synthetic OOD samples.")
else:
    print(
        "Average confidence did not decrease on synthetic OOD samples. "
        "This does not guarantee reliable OOD predictions."
    )

print(
    "Synthetic OOD accuracy cannot be measured reliably without "
    "ground-truth labels for the modified samples."
)
```
